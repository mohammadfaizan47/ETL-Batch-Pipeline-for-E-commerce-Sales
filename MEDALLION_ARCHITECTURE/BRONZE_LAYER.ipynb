```python
# Looking out data and seeing problems in data for further clearning process
from pyspark.sql import *

orders_path = '/Volumes/ecom-medallion-pipeline/raw/mobile_orders/flipkart_mobiles.csv'

df = spark.read.csv(orders_path, header=True, inferSchema=True)
display(df)
```

```python
# adding metadata for detecting errors or bugs in our data
from pyspark.sql.functions import current_timestamp, current_date, lit, col
import uuid

df = df.withColumn("ingestion_time", current_timestamp()) \
       .withColumn("load_date", current_date()) \
       .withColumn("batch_id", lit(str(uuid.uuid4()))) \
       .withColumn("source_file", col("_metadata.file_path"))   # Unity Catalog

display(df)
```

```python
# writing data to bronze layer
bronze_path = "/Volumes/ecom-medallion-pipeline/bronze/mobile_orders"

df.write.format("parquet") \
    .mode("overwrite") \
    .partitionBy("load_date") \
    .save(bronze_path)

display(df)
```
